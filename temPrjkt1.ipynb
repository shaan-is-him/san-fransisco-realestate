```python
# We only keep rows that have a meaningful cost
# very low or zero costs are usually data errors 
new_df = new_df[new_df["Revised Cost"] > 1000].copy()

# Fill missing numeric values with the median
# Median is better than mean because cost and size data are skewed
numeric_cols = [
    "Number of Existing Stories", "Number of Proposed Stories",
    "Existing Units", "Proposed Units", "Plansets",
    "Existing Construction Type", "Proposed Construction Type",
    "Supervisor District", "Zipcode", "Estimated Cost", "Revised Cost"
]

for col in numeric_cols:
    if col in new_df.columns:
        new_df[col] = new_df[col].fillna(new_df[col].median())

# Fill missing categorical values with "Unknown"
cat_cols = [
    "Permit Type Definition",
    "Existing Use", "Proposed Use",
    "Existing Construction Type Description",
    "Proposed Construction Type Description"
]

for col in cat_cols:
    if col in new_df.columns:
        new_df[col] = new_df[col].fillna("Unknown")

print("Shape after cleaning:", new_df.shape)
```

```python
# IMPORTANT: We deliberately EXCLUDE "Neighborhoods - Analysis Boundaries"
# Reason: We want the algorithm to find groups based on the NATURE of the project

features = [
    # Numeric features 
    "Number of Existing Stories",
    "Number of Proposed Stories",
    "Existing Units",
    "Proposed Units",
    "Plansets",
    "Estimated Cost",
    "Revised Cost",
    
    # Construction type codes numeric but categorical in nature
    "Existing Construction Type",
    "Proposed Construction Type",
    
    # Categorical features 
    "Permit Type Definition",
    "Existing Use",
    "Proposed Use",
    "Existing Construction Type Description",
    "Proposed Construction Type Description"
]

# Keeping only the columns we need + the neighborhood column for later analysis
df_cluster = new_df[features + ["Neighborhoods - Analysis Boundaries"]].copy()
df_cluster = df_cluster.dropna()

print("Shape used for clustering:", df_cluster.shape)
```

```python
X = df_cluster[features] # features used for clustering
neighborhoods = df_cluster["Neighborhoods - Analysis Boundaries"] # kept only for anaylisis
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder


num_features = [
    "Number of Existing Stories", "Number of Proposed Stories",
    "Existing Units", "Proposed Units", "Plansets",
    "Revised Cost",
    "Existing Construction Type", "Proposed Construction Type"
]

cat_features = [
    "Permit Type Definition",
    "Existing Use", "Proposed Use",
    "Existing Construction Type Description",
    "Proposed Construction Type Description"
]

# StandardScaler puts all numeric columns on the same scale
# OneHotEncoder converts categories into 0/1 columns

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), num_features),
        ("cat", OneHotEncoder(sparse_output=False), cat_features)
    ]
)
```

```python
# A common good choice for this kind of data is between 5 and 8.
# You can change this value after looking at the plots.
from sklearn.pipeline import Pipeline
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
best_k = 3
# pipeline lets us sequentially apply series of transformation insuring it stays same throughout the data
pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("cluster", KMeans(n_clusters=3, random_state=42, n_init=10))
])

pipeline.fit(X)

# getting the labels
cluster_labels = pipeline.predict(X)

# Adding the cluster label back to the dataframe
df_cluster["Cluster"] = cluster_labels
print(df_cluster["Cluster"].value_counts().sort_index())
```

```python
print("CLUSTER vs NEIGHBORHOOD ")

ct = pd.crosstab(
    df_cluster["Cluster"],
    df_cluster["Neighborhoods - Analysis Boundaries"],
    normalize="index" # show percentages within each cluster
)

# Show the top 5 neighborhoods for each cluster
for cluster_id in sorted(df_cluster["Cluster"].unique()):
    print(f"\n----- Cluster {cluster_id} -----")
    top_neigh = ct.loc[cluster_id].sort_values(ascending=False).head(5)
    for neigh, pct in top_neigh.items():
        print(f"{neigh:45s} {pct*100:5.1f}%")
```

```python
print("CLUSTER PROFILES average values")

profile = df_cluster.groupby("Cluster").agg({
    "Revised Cost": "mean",
    "Estimated Cost": "mean",
    "Number of Proposed Stories": "mean",
    "Proposed Units": "mean",
    "Number of Existing Stories": "mean",
    "Existing Units": "mean",
    "Plansets": "mean"
}).round(1)

print(profile)
```

```python
# PCA reduces the high-dimensional data to 2D so we can plot it
from sklearn.decomposition import PCA

X_transformed = pipeline.named_steps["preprocessor"].transform(X)
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_transformed)

plt.figure(figsize=(20, 7))
scatter = plt.scatter(
    X_pca[:, 0], X_pca[:, 1],
    c=cluster_labels,
    cmap="tab10",
    s=8,
    alpha=0.6
)
plt.colorbar(scatter, label="Cluster")
plt.title("Clusters visualized with PCA")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.show()
```
